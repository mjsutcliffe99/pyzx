```python
def gen_graph(do_param):
    q,d,seed = 10,300,42
    g = zx.generate.cliffordT(q,d,seed=seed)
    g.apply_state("0110100000")
    g.apply_effect("0"*q)

    if (do_param):
        g.set_phase(0,zx.symbolic.new_var('a',True))
        g.set_phase(1,zx.symbolic.new_var('b',True))
        g.set_phase(2,zx.symbolic.new_var('c',True))
        g.set_phase(3,zx.symbolic.new_var('d',True))
        g.set_phase(4,zx.symbolic.new_var('e',True))
        g.set_phase(5,zx.symbolic.new_var('f',True))
        g.set_phase(6,zx.symbolic.new_var('g',True))
        g.set_phase(7,zx.symbolic.new_var('h',True))
        g.set_phase(8,zx.symbolic.new_var('i',True))
        g.set_phase(9,zx.symbolic.new_var('j',True))

    return g
```

## Parallel Eval

```python
import math
from dataclasses import dataclass
from fractions import Fraction
from functools import partial

import numpy as np
import jax
import jax.numpy as jnp

from pyzx.graph.scalar import Scalar, cexp, phase_pair_value
from pyzx.symbolic import Poly

jax.config.update("jax_enable_x64", True)


def _poly_parts(p) -> tuple[Fraction, dict[tuple[str, ...], Fraction]]:
    """Split a phase into (constant, {monomial: coefficient}), a monomial being a sorted tuple of
    Boolean variable names (x^k = x for a Boolean x)."""
    if not isinstance(p, Poly):
        return Fraction(p), {}
    const, mons = Fraction(0), {}
    for c, t in p.terms:
        if isinstance(c, complex):
            raise ValueError(f"Complex coefficient in phase {p}")
        if not all(v.is_bool for v, _ in t.vars):
            raise ValueError(f"Phase {p} has a non-Boolean variable")
        names = tuple(sorted({v.name for v, _ in t.vars}))
        if names:
            mons[names] = mons.get(names, 0) + Fraction(c)
        else:
            const += Fraction(c)
    return const, {m: c for m, c in mons.items() if c % 2 != 0}


def _is_concrete(p) -> bool:
    return not _poly_parts(p)[1]


@dataclass
class CompiledScalarSum:
    """A sum of parameterised scalars compiled into integer arrays (cf. arXiv:2403.06777, Sec. 3.3).

    Every phase (in units of pi) is stored as an integer row ``A`` and offset ``b`` in units of
    pi/N, so that its value for a bitstring is ``(A @ m + b) mod 2N``, where ``m`` is the 0/1 vector
    of monomials (products of variables) present. For the paper's polarised phases ``alpha + pi*XOR(P)``,
    ``A`` is just 4x the indicator vector of ``P`` (with N = 4), and the matmul mod 8 is the XOR.

    Term ``t`` evaluates to
        const[t] * e^{i pi phase} * prod_k (1 + e^{i pi node_k}) * prod_k e^{i pi c_k r(p1_k) r(p2_k)}
                 * sum_k coeff_k e^{i pi sum_k}
    with ``r`` reducing into [0, 2). Subterms that are concrete are folded into ``const`` at compile
    time. Each subterm type is padded to the max count over terms: padded nodes are masked out,
    padded pairs and sum entries have coefficient 0.
    """
    var_names: list[str]
    monomials: list[tuple[str, ...]]
    N: int  # phases are integers mod 2N, in units of pi/N
    Q: int  # pair coefficients are integers in units of 1/Q
    arrays: dict[str, np.ndarray]

    def evaluate(self, bitstrings) -> np.ndarray | complex:
        """Evaluate the scalar sum for one bitstring or a batch of them.

        ``bitstrings`` can be a string like ``"01101"`` or a dict ``{name: bit}`` (one
        assignment), or a list of those / a (B, n) 0/1 array (a batch, evaluated in parallel)."""
        x, single = self._to_array(bitstrings)
        out = _evaluate_batch(self.arrays, jnp.asarray(x), self.N, self.Q)
        return complex(out[0]) if single else np.asarray(out)

    def _to_array(self, bitstrings) -> tuple[np.ndarray, bool]:
        def row(b):
            if isinstance(b, dict):
                return [int(b[v]) for v in self.var_names]
            if isinstance(b, str):
                b = [int(c) for c in b]
            if len(b) != len(self.var_names):
                raise ValueError(f"Expected {len(self.var_names)} bits for variables {self.var_names}")
            return list(b)
        if isinstance(bitstrings, (str, dict)):
            return np.array([row(bitstrings)], dtype=np.int64), True
        if isinstance(bitstrings, np.ndarray) and bitstrings.ndim == 2:
            return bitstrings.astype(np.int64), False
        return np.array([row(b) for b in bitstrings], dtype=np.int64).reshape(-1, len(self.var_names)), False


def compile_scalar_sum(scalars: list[Scalar], var_names: list[str] | None = None) -> CompiledScalarSum:
    """Compile a sum of parameterised scalars (e.g. the output of ``simulate`` on a parameterised
    diagram) into a :class:`CompiledScalarSum`. ``var_names`` fixes the bit order (default: sorted)."""
    if isinstance(scalars, Scalar):
        scalars = [scalars]

    # Per term: constant, and the remaining symbolic subterms.
    terms = []
    for s in scalars:
        if s.is_unknown:
            raise ValueError("Cannot compile an unknown scalar")
        if s.is_zero:
            continue
        const = complex(s.floatfactor) * math.sqrt(2) ** s.power2
        nodes = []
        for n in s.phasenodes:
            if _is_concrete(n):
                const *= 1 + cexp(_poly_parts(n)[0])
            else:
                nodes.append(n)
        pairs = []
        for c, p1, p2 in s.phasepairs:
            if _is_concrete(p1) and _is_concrete(p2):
                const *= cexp(phase_pair_value(c, _poly_parts(p1)[0], _poly_parts(p2)[0]))
            else:
                pairs.append((Fraction(c), p1, p2))
        sop = list(s.sum_of_phases.items()) or [(0, 1)]
        if all(_is_concrete(p) for p, _ in sop):
            const *= sum(c * cexp(_poly_parts(p)[0]) for p, c in sop)
            sop = [(0, 1)]
        if const != 0:
            terms.append((const, s.phase, nodes, pairs, sop))

    # Variables, monomials and the common phase unit pi/N.
    polys = [p for _, ph, nodes, pairs, sop in terms
             for p in [ph, *nodes, *(q for _, p1, p2 in pairs for q in (p1, p2)), *(p for p, _ in sop)]]
    parts = [_poly_parts(p) for p in polys]
    monomials = sorted({m for _, mons in parts for m in mons}, key=lambda m: (len(m), m))
    found = sorted({v for m in monomials for v in m})
    if var_names is None:
        var_names = found
    elif not set(found) <= set(var_names):
        raise ValueError(f"Variables {set(found) - set(var_names)} missing from var_names")
    N = math.lcm(1, *(c.denominator for const, mons in parts for c in [const, *mons.values()]))
    Q = math.lcm(1, *(c.denominator for _, _, _, pairs, _ in terms for c, _, _ in pairs))

    mon_index = {m: i for i, m in enumerate(monomials)}
    T, M = len(terms), len(monomials)
    Kn = max([len(t[2]) for t in terms], default=0)
    Kp = max([len(t[3]) for t in terms], default=0)
    Ks = max([len(t[4]) for t in terms], default=1)

    def encode(p, A, b, idx):
        const, mons = _poly_parts(p)
        b[idx] = int(const * N) % (2 * N)
        for m, c in mons.items():
            A[idx + (mon_index[m],)] = int(c * N) % (2 * N)

    arr = {
        "mono": np.zeros((M, len(var_names)), np.int64),
        "mono_deg": np.array([len(m) for m in monomials], np.int64),
        "const": np.zeros(T, np.complex128),
        "phase_A": np.zeros((T, M), np.int64), "phase_b": np.zeros(T, np.int64),
        "node_A": np.zeros((T, Kn, M), np.int64), "node_b": np.zeros((T, Kn), np.int64),
        "node_mask": np.zeros((T, Kn), bool),
        "pair_A1": np.zeros((T, Kp, M), np.int64), "pair_b1": np.zeros((T, Kp), np.int64),
        "pair_A2": np.zeros((T, Kp, M), np.int64), "pair_b2": np.zeros((T, Kp), np.int64),
        "pair_c": np.zeros((T, Kp), np.int64),
        "sum_A": np.zeros((T, Ks, M), np.int64), "sum_b": np.zeros((T, Ks), np.int64),
        "sum_coeff": np.zeros((T, Ks), np.int64),
    }
    for i, m in enumerate(monomials):
        for v in m:
            arr["mono"][i, var_names.index(v)] = 1
    for t, (const, ph, nodes, pairs, sop) in enumerate(terms):
        arr["const"][t] = const
        encode(ph, arr["phase_A"], arr["phase_b"], (t,))
        for k, n in enumerate(nodes):
            encode(n, arr["node_A"], arr["node_b"], (t, k))
            arr["node_mask"][t, k] = True
        for k, (c, p1, p2) in enumerate(pairs):
            encode(p1, arr["pair_A1"], arr["pair_b1"], (t, k))
            encode(p2, arr["pair_A2"], arr["pair_b2"], (t, k))
            arr["pair_c"][t, k] = int(c * Q)
        for k, (p, c) in enumerate(sop):
            encode(p, arr["sum_A"], arr["sum_b"], (t, k))
            arr["sum_coeff"][t, k] = c
    return CompiledScalarSum(list(var_names), monomials, N, Q, arr)


@partial(jax.jit, static_argnames=("N", "Q"))
def _evaluate_batch(arr: dict, x: jax.Array, N: int, Q: int) -> jax.Array:
    """Evaluate the compiled sum for a (B, n) batch of bitstrings. Vectorised over bitstrings,
    terms and subterms, so on a GPU every (bitstring, term, subterm) is handled in parallel."""
    # Lookup tables e^{i pi k/N} and, for pairs, e^{i pi k/(Q N^2)} (paper: Lemma 7 tables).
    E = jnp.exp(1j * jnp.pi * jnp.arange(2 * N) / N)
    E_pair = jnp.exp(1j * jnp.pi * jnp.arange(2 * Q * N * N) / (Q * N * N))

    def one(xb):
        m = (arr["mono"] @ xb == arr["mono_deg"]).astype(jnp.int64)  # which monomials are 1
        phase = lambda A, b: (A @ m + b) % (2 * N)  # in units of pi/N, reduced into [0, 2N)

        val = arr["const"] * E[phase(arr["phase_A"], arr["phase_b"])]
        nodes = jnp.where(arr["node_mask"], 1 + E[phase(arr["node_A"], arr["node_b"])], 1)
        u1, u2 = phase(arr["pair_A1"], arr["pair_b1"]), phase(arr["pair_A2"], arr["pair_b2"])
        pairs = E_pair[(arr["pair_c"] * u1 * u2) % (2 * Q * N * N)]
        sums = (arr["sum_coeff"] * E[phase(arr["sum_A"], arr["sum_b"])]).sum(axis=1)
        val = val * nodes.prod(axis=1) * pairs.prod(axis=1) * sums
        return val.sum()  # sum over terms (a parallel tree reduction, cf. paper App. E)

    return jax.vmap(one)(x)
```

```python
import itertools

compiled = compile_scalar_sum(res)
print("variables:", compiled.var_names, "| monomials:", compiled.monomials)
print({k: v.shape for k, v in compiled.arrays.items()})

# One bitstring (same assignment as above): should match `total`
print(compiled.evaluate(assignment), "vs", total)

# All 2^n bitstrings at once, evaluated in parallel
bits = np.array(list(itertools.product([0, 1], repeat=len(compiled.var_names))))
amps = compiled.evaluate(bits)
for b, amp in zip(bits, amps):
    print("".join(map(str, b)), amp)
```
