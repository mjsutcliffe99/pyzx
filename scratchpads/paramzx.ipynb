```python
import pyzx as zx
```

```python
def gen_graph(do_param):
    q,d,seed = 5,200,42
    g = zx.generate.cliffordT(q,d,seed=seed)
    g.apply_state("01101")
    g.apply_effect("0"*q)

    if (do_param):
        g.set_phase(0,zx.symbolic.new_var('a',True))
        g.set_phase(1,zx.symbolic.new_var('b',True))
        g.set_phase(2,zx.symbolic.new_var('c',True))
        g.set_phase(3,zx.symbolic.new_var('d',True))
        g.set_phase(4,zx.symbolic.new_var('e',True))

    return g
```

## Non-param

```python
g = gen_graph(do_param=False)
zx.draw(g,scale=20)
res = zx.simulation.simulate(zx.simulation.Strategy.BSS,g)
res
```

```python
g = gen_graph(do_param=False)
zx.draw(g,scale=20)
res = zx.simulation.simulate(zx.simulation.Strategy.MAGIC_CAT,g)
res
```

```python
#(-0.16115169529663687-0.07276334764831836j)
```

## Param

```python
g = gen_graph(do_param=True)
zx.draw(g,scale=20)
res = zx.simulation.simulate(zx.simulation.Strategy.MAGIC_CAT,g)
res
```

```python
assignment = {'a': 0, 'b': 1, 'c': 1, 'd': 0, 'e': 1}

total = 0
for i, scalar in enumerate(res):
    var_map = {v: assignment[v.name] for v in scalar.free_vars()}
    val = scalar.substitute_variables(var_map).to_number()
    print(f"term {i}: {val}")
    total += val

print()
print("total:", total)
```
